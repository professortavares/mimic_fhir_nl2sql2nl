# Notebook 1 — Teste de conexão com a OpenAI via LangChain

Este notebook demonstra como usar LangChain para fazer uma chamada simples ao modelo GPT-4o-mini da OpenAI.
Verifica que:
1. A chave de API está configurada corretamente
2. LangChain consegue criar uma instância do modelo
3. O modelo consegue processar uma pergunta e retornar uma resposta

```python
from nl2sql2nl.llm import criar_modelo, chamar_modelo

print("Módulos importados com sucesso!")
```

## Criando o modelo

```python
# Cria uma instância do modelo com parâmetros padrão
llm = criar_modelo()

print(f"Modelo criado: {llm.model_name}")
print(f"Temperatura: {llm.temperature}")
print(f"Tipo: {type(llm).__name__}")
```

## Fazendo uma pergunta simples

```python
# Faz uma pergunta usando o wrapper chamar_modelo
pergunta = "Olá! Em uma frase, o que é NL2SQL?"

print(f"Pergunta: {pergunta}")
print("\nAguardando resposta do modelo...\n")

resposta = chamar_modelo(pergunta)

print(f"Resposta: {resposta}")
```

## Chamada direta via LangChain (sem wrapper)

```python
from langchain_core.messages import HumanMessage

# Cria mensagem de usuário e invoca o modelo diretamente
mensagens = [HumanMessage(content="Explique SQL em uma frase.")]

print("Invocando modelo com HumanMessage...\n")

resposta_direta = llm.invoke(mensagens)

print(f"Resposta: {resposta_direta.content}")
```
